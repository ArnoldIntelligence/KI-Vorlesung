# 06-Convolutional Neural Networks

CNNs sind neuronale Netze, die mithilfe von Filtern automatisch Merkmale aus Daten – besonders aus Bildern – extrahieren. Durch Gewichtsteilung und lokale Filter benötigen sie deutlich weniger Parameter als vollständig verbundene Netze und können so effizient komplexe Muster erkennen.

% 1D-Faltung (diskret)
$(y * w)[n] = \sum_{k=-\infty}^{\infty} x[k] \, w[n-k]$

% 1D-Kreuzkorrelation (typisch in CNNs)
$y[n] = \sum_{k=-\infty}^{\infty} x[n+k] \, w[k]$

% 2D-Faltung
$Y[i,j] = \sum_{u=0}^{m-1}\sum_{v=0}^{n-1} X[i-u, j-v] \, K[u,v]$

% 2D-Kreuzkorrelation (in CNNs üblich)
$Y[i,j] = \sum_{u=0}^{m-1}\sum_{v=0}^{n-1} X[i+u, j+v] \, K[u,v]$

% Mit Bias-Term
$Y[i,j] = \sum_{u=0}^{m-1}\sum_{v=0}^{n-1} X[i+u, j+v] \, K[u,v] + b$

% Mehrkanal-Faltung (z.B. RGB)
$Y^{(r)}[i,j] =\sum_{c=1}^{C}\sum_{u=0}^{m-1}\sum_{v=0}^{n-1} X[i+u, j+v, c] \, K^{(r)}[u,v,c] + b^{(r)}$

% Mit Stride s
$Y[i,j] = \sum_{u=0}^{m-1}\sum_{v=0}^{n-1} X[i \cdot s + u, j \cdot s + v] \, K[u,v] + b$
